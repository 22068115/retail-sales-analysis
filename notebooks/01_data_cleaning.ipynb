```python
sales["calculated_value"] = sales["quantity"] * sales["unit_price"]

sales[
    ["quantity", "unit_price", "total_value", "calculated_value", "discount_pct"]
].head(10)
```

```python
(sales["total_value"] - sales["calculated_value"]).abs().describe()
```

```python
# Remove temporary validation column
sales.drop(columns=["calculated_value"], inplace=True)

# Convert date
sales["date"] = pd.to_datetime(sales["date"])

# Fix customer ID datatype while keeping missing customers
sales["customer_id"] = sales["customer_id"].astype("Int64")

# Remove duplicate rows
sales = sales.drop_duplicates()

# Calculate revenue after discount
sales["net_sales"] = (
    sales["total_value"] * (1 - sales["discount_pct"] / 100)
)

# Create useful date columns
sales["year"] = sales["date"].dt.year
sales["month"] = sales["date"].dt.month
sales["month_name"] = sales["date"].dt.month_name()
sales["day_of_week"] = sales["date"].dt.day_name()

# Check result
print("Rows:", len(sales))
print("Duplicates:", sales.duplicated().sum())
print("\nMissing values:")
print(sales.isnull().sum())
```

```python
sales.to_csv(
    "../data/cleaned/bm_sales_cleaned.csv",
    index=False
)

print("Cleaned sales data saved successfully!")
```
